## **Step 8: Remove exact duplicate rows in the data**

### **`drop_duplicates()`:**

* **`df.drop_duplicates()`:** removes **duplicate rows**, not individual repeated values.

* By default, Pandas compares **all columns** to determine whether two rows are duplicates.

* If two rows are completely identical, Pandas keeps the first occurrence and removes the others.

* Repeated values in a single column are **not necessarily duplicates**. 

* For example, the same `CustomerID` or `Product` can appear in multiple valid orders. 

* **To check duplicated rows before removing them:** `df[df.duplicated()]`

* **To count the number of duplicated rows:** `df.duplicated().sum()`

* **To remove duplicated rows:** `df = df.drop_duplicates()`

**Important:** `drop_duplicates()` removes only rows that are duplicated based on the selected columns. It does not remove every repeated value in the dataset.

````python
df = df.drop_duplicates()
df.head(10)
````

## **Step 9: Clean text columns**

* #### **`.value_counts()`**

    * **`.value_counts()`:** counts the **number of occurrences of each unique value** in a Series or column.
    
    * It is useful for understanding the **distribution of categorical data**.
    
    * By default, the results are sorted from the **most frequent value to the least frequent value**.

    **Common use:** Checking the distribution of categories and detecting unexpected or inconsistent values during data cleaning.

* #### **`str.title()`**

    * **`.str.title()`:** converts a string so that the **first letter of each word is capitalized** and the remaining letters are lowercase.

    * It is useful for **standardizing text formatting** during data cleaning.

    This helps make categorical values consistent and easier to analyze.

* #### **Why do we use `.str`?**

    * **A single Python string can directly use string methods such as `.title()` and `.strip()`:** 
    `text.title()` 
    `text.strip()`

    * However, a Pandas column is a **Series**, not a single string.
    
    * To apply string methods to every value in a Pandas Series, we use the `.str` accessor.
    `df["Region"].str.title()`
    `df["Region"].str.strip()`

    * `.str` tells Pandas to apply the following **string operation element-wise** to the values in the Series.

    **In short:**

    ```text
    Single String  →  .title() / .strip()
    Pandas Series  →  .str.title() / .str.strip()
    ```

* #### **`dropna=False in value_counts()`:**
    * `df["Category"].value_counts(dropna=False)`
    
    * **`value_counts()`:** counts how many times each unique value appears
    
    * By default, `value_counts()` ignores missing values (NaN).

    * `dropna=False` tells Pandas to include NaN values in the count.

    * **Example:**

        ```text python
        A      3

        B      2

        NaN    2

        So:

        value_counts()                  # ignores NaN
        value_counts(dropna=False)      # includes NaN
        ```

    **💡 Useful for EDA:** It helps you see both the distribution of values and the number of missing values in a column.

## **Step 11: Fix dates**

#### **Converting `OrderDate` to Datetime**

* **The following code converts the `OrderDate` column from strings or mixed date formats into Pandas datetime values:**

    ```python
    df["OrderDate"] = pd.to_datetime(
        df["OrderDate"],
        format="mixed",
        errors="coerce"
    )
    ```

* ##### **Step-by-step:**

    * **`df["OrderDate"]`:** Selects the `OrderDate` column from the DataFrame.

    * **`pd.to_datetime()`:** Converts the values into Pandas datetime objects.

    * **`format="mixed"`:** Allows the column to contain dates in different formats.

    * **`errors="coerce"`:** Converts invalid or unrecognized date values to `NaT` instead of raising an error.

    * **`df["OrderDate"] =`:** Stores the converted values back into the `OrderDate` column.


* #####  **Why is this useful?**

    Converting the column to datetime makes it easier to perform date-related operations such as sorting, filtering, extracting the year/month/day, and calculating differences between dates.

#### **8- How do customer types compare in terms of total revenue, number of orders, and average order value?**

* ##### **`GroupBy` with `agg()` in Pandas**

    * **`groupby()`:** is used to split the DataFrame into groups based on one or more columns.

    * **`agg()`:** (aggregation) is used to apply one or more aggregate functions to each group.

* ##### **Basic Syntax:**

    ```python
    df.groupby("group_column").agg(
        NewColumnName=("column_to_aggregate", "function")
    )
    ```

    ###### **Example:**

    ```python
    customer_type_summary = df.groupby("CustomerType").agg(
        TotalRevenue=("Revenue", "sum")
    )
    ```

    * **Explanation:**

        * **`groupby("CustomerType")`** → Groups the data by customer type.

        * **`("Revenue", "sum")`** → Calculates the sum of `Revenue` for each group.

        * **`TotalRevenue`** → The name given to the resulting column.

        The result will look like:

        ```text
                    TotalRevenue
        CustomerType
        New                 15000
        Returning           32000
        ```

* ##### **Multiple Aggregations:**

    **We can apply multiple aggregate functions at the same time:**

    ```python
    customer_type_summary = df.groupby("CustomerType").agg(
        TotalRevenue=("Revenue", "sum"),
        AverageRevenue=("Revenue", "mean"),
        MaximumRevenue=("Revenue", "max")
    )
    ```

    This allows us to calculate multiple summary statistics for each customer type in one operation.

* ##### **Common aggregate functions:**

    * `sum()` → Total value
    * `mean()` → Average value
    * `min()` → Minimum value
    * `max()` → Maximum value
    * `count()` → Number of non-null values
    * `nunique()` → Number of unique values
    * `median()` → Median value

````python
customer_type_summary = df.groupby("CustomerType").agg(
    TotalRevenue= ("Revenue", "sum"),
    Orders= ("OrderID", "nunique"),
    AvgOrderValue= ("Revenue", "mean")
).sort_values("TotalRevenue", ascending=False)

customer_type_summary
````
